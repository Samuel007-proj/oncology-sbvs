```python
from rdkit import Chem
from rdkit.Chem import Draw
from pathlib import Path
import pandas as pd
```

```python

mols = [
    {   "Mol_name": "T1_MYF_03_176",
        "SMILES" : "FC(=C)C(=O)N1C[C@@H](NC2=NC=CC=N2)[C@@H](C1)OCC1=CC=C(C=C1)C(F)(F)F"
    },
    {
        "Mol_name": "P1_Compd_31", 
        "SMILES": "C[n]1ncc2c1c(cc(C(N(Cc(nc1)ccc1C#CC1=CCN(C)CC1)C1CC1)=O)c(F)c1)c1nc2N"
    }
]
```

```python
path= Path('./')

for file in path.glob('*.sdf'):
    print(file)
    sdf = next(Chem.SDMolSupplier(file))
    smiles = Chem.MolToSmiles(sdf)
    
    mols.append({
        "Mol_name" : file.stem,
        "SMILES": smiles
        })

    print(file.stem, smiles, sep=' ==> ')
```

```python
df = pd.DataFrame(mols)

df
```

```python

df.to_csv('cohort_1.csv', index=False)
```

### Draw

```python
img = Draw.MolsToImage(
    mols = [Chem.MolFromSmiles(smi) for smi in df['SMILES']],
    molsPerRow=3,
    subImgSize=(200,200),
    legends=[name for name in df['Mol']]
)

img.show()
```
